```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer

# Sample text corpus
text = "deep learning is interesting deep learning uses neural networks"

# Tokenize words
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

words = tokenizer.word_index
vocab_size = len(words) + 1

# Create CBOW context-target pairs
sequence = tokenizer.texts_to_sequences([text])[0]
X, y = [], []

window = 2

for i in range(window, len(sequence) - window):
    context = sequence[i-window:i] + sequence[i+1:i+window+1]
    target = sequence[i]
    X.append(context)
    y.append(target)

X = np.array(X)
y = np.array(y)

# Build CBOW model
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(vocab_size, 10, input_length=4),
    tf.keras.layers.Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    tf.keras.layers.Dense(vocab_size, activation='softmax')
])

# Compile and train
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=0)

# Display predictions
reverse_words = {v: k for k, v in words.items()}

predictions = model.predict(X)

print("CBOW Sample Predictions:")

for i in range(min(5, len(X))):
    predicted_word = reverse_words[np.argmax(predictions[i])]
    actual_word = reverse_words[y[i]]
    print("Actual:", actual_word, "| Predicted:", predicted_word)
```
